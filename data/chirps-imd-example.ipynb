# Do not join: CHIRPS grid vs IMD subdivision rainfall
Uses only the Python standard library.

CHIRPS is a 0.05° satellite–station **grid**. IMD rainfall is a **meteorological subdivision** (many districts). A lat/lon cell is not Pune district and is not Madhya Maharashtra.

This notebook shows there is no shared key and **refuses** the join. For India crop + rain use the IMD subdivision kit.


```python
import csv
from pathlib import Path

root = Path(".")
grid = list(csv.DictReader((root / "data/samples/chirps-grid-sample.csv").open()))
imd = list(csv.DictReader((root / "data/samples/india-rainfall-sample.csv").open()))
grid_keys = {(r["lon"], r["lat"], r["year"]) for r in grid}
imd_keys = {(r["SUBDIVISION"], r["YEAR"]) for r in imd}
print("chirps keys", sorted(grid_keys))
print("imd keys", sorted(imd_keys)[:3], "...")
print("shared keys", grid_keys & imd_keys)
assert not (grid_keys & imd_keys)
print("REFUSE join. Grid cell is not an IMD subdivision. Do not average CHIRPS to a district name.")
```
