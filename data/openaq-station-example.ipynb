# OpenAQ stations are not a city AQI
Uses only the Python standard library. The Pune sample is three stations on one day.

**Do not join** these hours to World Bank `EN.ATM.PM25.MC.M3` (national annual mean).
**Do not average** stations into one city number. Report min / median / max at **station** grain.


```python
import csv, statistics
from collections import defaultdict
from pathlib import Path

rows = list(csv.DictReader((Path(".") / "data/samples/openaq-pune-sample.csv").open()))
by_station = defaultdict(list)
for row in rows:
    by_station[row["station_id"]].append(float(row["value"]))
means = {sid: statistics.mean(vals) for sid, vals in by_station.items()}
spread = max(means.values()) / min(means.values())
assert len(means) >= 3, "need multiple stations"
assert spread >= 2, f"intra-city spread {spread:.1f}x is too small to demonstrate the trap"
print("station mean_pm25")
for sid, mean in sorted(means.items()):
    print(sid, round(mean, 1))
print(f"city={rows[0]['city']} stations={len(means)} min={min(means.values()):.1f} median={statistics.median(means.values()):.1f} max={max(means.values()):.1f} spread={spread:.1f}x")
print("REFUSE city-average AQI. REFUSE join to national annual PM2.5. Result grain is station.")
```
